```python
import pandas as pd
```

```python
dataset = pd.read_csv("marks.csv")
dataset
```

```python
independent = dataset[["Hours_Studied"]]
dependent = dataset[["Exam_Marks"]]

print(independent,dependent)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(x_train,y_train)

weight = regressor.coef_
bias = regressor.intercept_
```

```python
weight , bias
```

```python
regressor.predict([[3]])
```

```python
from sklearn.metrics import r2_score
y_pred = regressor.predict(x_test)
r_score = r2_score(y_test,y_pred)
r_score
```

```python
import matplotlib.pyplot as plt
X = independent
y = dependent
plt.scatter(X, y)
plt.plot(X, regressor.predict(X))
plt.show()
```
