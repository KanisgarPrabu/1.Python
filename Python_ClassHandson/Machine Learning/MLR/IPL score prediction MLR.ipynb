```python
"""Problem Statement

Predict the Final IPL Score using:

Score after 10 overs
Wickets lost
Run rate
Last 5 overs momentum
Venue
Pitch type
Batting team
Dew presence
Dataset Details
5000 rows
Numerical + categorical features
Realistic cricket-style scoring patterns
Perfect for:
Multiple Linear Regression
One Hot Encoding
Feature Engineering
Train/Test Split
R² Evaluation"""
```

```python
import pandas as pd
```

```python
dataset = pd.read_csv("ipl_multiple_linear_regression_dataset.csv")
dataset[['Venue']]
```

```python
dataset = pd.get_dummies(dataset,dtype = int,drop_first=False)
```

```python
dataset.columns
```

```python
independent = dataset[['ScoreAfter10Overs', 'WicketsLost', 'Last5OversRuns',
       'DewPresent', 'Venue_Bangalore', 'Venue_Chennai',
       'Venue_Delhi', 'Venue_Kolkata', 'Venue_Mumbai', 'PitchType_Flat',
       'PitchType_Green', 'PitchType_Slow', 'PitchType_Turning',
       'BattingTeam_CSK', 'BattingTeam_DC', 'BattingTeam_KKR',
       'BattingTeam_MI', 'BattingTeam_RCB', 'BattingTeam_SRH']]
dependent = dataset [['FinalScore']]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train,y_train)
```

```python
y_pred = regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
r_score
```
