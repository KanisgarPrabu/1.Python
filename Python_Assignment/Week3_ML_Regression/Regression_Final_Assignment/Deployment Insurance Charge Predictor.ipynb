```python
import pickle

model = pickle.load(open('insurance_charge_predictor.sav','rb'))
age = int(input("Enter the age of the insurer"))
gender = int(input("Enter 0 for Female, 1 for Male"))
bmi = float(input("Enter the BMI of the insurer"))
children = int(input("Enter number of children for the insurer"))
smoker = int(input("Enter 1 for Smoker , 0 for Non-Smoker"))
```

```python
model.predict([[age,gender,bmi,children,smoker]])
```
